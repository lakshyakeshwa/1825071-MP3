```python
def rk4(s):
    h = 0.1
    y = 0
    z = s
    x = 0

    while x < 1:
        y += h*z
        x += h

    return y

a = 0
b = 2

for i in range(20):
    c = b - (rk4(b)-1)*(b-a)/(rk4(b)-rk4(a))

    a = b
    b = c

print("Required slope:", c)
print("Final value:", rk4(c))
```
