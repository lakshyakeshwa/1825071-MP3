```python

import cmath
import math

z = complex(4, 4)
n = 3

r = abs(z)
theta = cmath.phase(z)

for k in range(n):
    angle = (theta + 2*math.pi*k)/n
    root = r**(1/n) * cmath.exp(1j*angle)

    print("Root:", root)
    print("Cartesian form:", root.real, "+", root.imag, "i")
    print("Polar form:", abs(root), "angle", angle)
    print()
```
