```python
import numpy as np
import matplotlib.pyplot as plt

a = 0
b = 1
n = 10

h = (b-a)/n

x = np.linspace(a, b, n+1)

y = np.zeros(n+1)

y[0] = 0
y[n] = 1

A = np.zeros((n-1, n-1))
B = np.zeros(n-1)

for i in range(n-1):
    A[i, i] = -2

    if i > 0:
        A[i, i-1] = 1

    if i < n-2:
        A[i, i+1] = 1

B[-1] = -1

y[1:n] = np.linalg.solve(A, B)

print("x values:", x)
print("y values:", y)

plt.plot(x, y, 'o-')
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```
