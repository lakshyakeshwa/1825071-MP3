```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)

X, Y = np.meshgrid(x, y)

Z = X + 1j*Y
W = Z**2

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter(Z.real, Z.imag)
plt.title("Z Plane")
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.grid()

plt.subplot(1, 2, 2)
plt.scatter(W.real, W.imag)
plt.title("W Plane (Z²)")
plt.xlabel("Real")
plt.ylabel("Imaginary")
plt.grid()

plt.show()
```
