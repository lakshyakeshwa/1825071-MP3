```python
def rk4(s):
    h = 0.1
    y = 0
    z = s
    x = 0

    while x < 1:
        y += h*z
        x += h

    return y

s = 0.5

for i in range(10):
    F = rk4(s) - 1
    dF = (rk4(s+0.001)-rk4(s))/0.001

    s = s - F/dF

print("Required slope:", s)
print("Final value:", rk4(s))
```
