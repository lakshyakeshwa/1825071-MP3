```python

import numpy as np
import matplotlib.pyplot as plt

z = np.array([2, 3])

translation = z + np.array([1, 2])
scaling = z * 2

theta = np.radians(45)
rotation = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
]) @ z

reflection = np.array([z[0], -z[1]])

print("Original:", z)
print("Translation:", translation)
print("Scaling:", scaling)
print("Rotation:", rotation)
print("Reflection:", reflection)

plt.quiver(0, 0, z[0], z[1], angles='xy',
           scale_units='xy', scale=1, label='Original')

plt.quiver(0, 0, rotation[0], rotation[1],
           angles='xy', scale_units='xy', scale=1,
           label='Rotation')

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.grid()
plt.legend()
plt.show()
```
