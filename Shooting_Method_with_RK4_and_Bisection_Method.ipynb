```python
def f(x, y, z):
    return z

def g(x, y, z):
    return 0

def rk4(s):
    h = 0.1
    x = 0
    y = 0
    z = s

    while x < 1:
        k1 = h*f(x, y, z)
        l1 = h*g(x, y, z)

        k2 = h*f(x+h/2, y+k1/2, z+l1/2)
        l2 = h*g(x+h/2, y+k1/2, z+l1/2)

        k3 = h*f(x+h/2, y+k2/2, z+l2/2)
        l3 = h*g(x+h/2, y+k2/2, z+l2/2)

        k4 = h*f(x+h, y+k3, z+l3)
        l4 = h*g(x+h, y+k3, z+l3)

        y += (k1+2*k2+2*k3+k4)/6
        z += (l1+2*l2+2*l3+l4)/6

        x += h

    return y

a = 0
b = 2

for i in range(30):
    c = (a+b)/2

    if (rk4(a)-1)*(rk4(c)-1) < 0:
        b = c
    else:
        a = c

print("Required initial slope:", c)
print("Final value:", rk4(c))
```
